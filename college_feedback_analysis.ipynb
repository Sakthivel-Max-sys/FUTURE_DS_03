```python
# -----------------------------------------------
#  College Event Feedback Analysis – Internship Project
# Using student_feedback.csv (Rating-based survey)
# -----------------------------------------------

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("student_feedback.csv")
print(" Dataset Loaded Successfully!")
print("Columns:", df.columns.tolist())

# -----------------------------------------------
# Step 1: Remove unwanted index column (if any)
# -----------------------------------------------
if 'Unnamed: 0' in df.columns:
    df.drop(columns=['Unnamed: 0'], inplace=True)

# -----------------------------------------------
# Step 2: Convert all rating columns to numeric
# -----------------------------------------------
for col in df.columns:
    if col != 'Student ID':
        df[col] = pd.to_numeric(df[col], errors='coerce')

# -----------------------------------------------
# Step 3: Basic summary
# -----------------------------------------------
print("\n Average Rating per Question:")
print(df.mean().round(2))

# -----------------------------------------------
# Step 4: Visualization - Average rating per question
# -----------------------------------------------
plt.figure(figsize=(10,5))
df.mean(numeric_only=True).plot(kind='bar', color='skyblue')
plt.title("Average Rating per Question")
plt.ylabel("Average Score")
plt.xticks(rotation=45, ha='right')
plt.show()

# -----------------------------------------------
# Step 5: Correlation Heatmap
# -----------------------------------------------
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation between Feedback Aspects")
plt.show()

# -----------------------------------------------
# Step 6: Distribution of one key question
# -----------------------------------------------
plt.figure(figsize=(6,4))
sns.histplot(df['Explains concepts in an understandable way'], bins=5, kde=True, color='green')
plt.title("Distribution - Explains concepts in an understandable way")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()

# -----------------------------------------------
# Step 7: Insights Summary
# -----------------------------------------------
print("\n Insights:")
print(" Highest-rated aspect:", df.mean().idxmax(), "→", round(df.mean().max(), 2))
print(" Lowest-rated aspect:", df.mean().idxmin(), "→", round(df.mean().min(), 2))
print(" Suggestion: Focus on improving lower-rated areas through interactive sessions or feedback follow-ups.")
```
